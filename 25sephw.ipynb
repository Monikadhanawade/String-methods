1. __sizeof__()

Use: Returns the memory size of the string object in bytes.

```python
s = "monika"

print(s.__sizeof__())
```

2. __subclasshook__()

Use: Helps Python determine relationships between classes and subclasses. It is mainly related to Python's class/inheritance system rather than normal string processing.

```python
print(str.__subclasshook__(str))
```

3. __capitalize__() 
Use: Converts the first character to uppercase.

```python
s = "monika"

print(s.capitalize())
```

4. __hash__()

Use: Returns the hash value of a string.

```python
s = "monika"

print(s.__hash__())
```

5. __get__()

str does not normally have a __get__() method.

```python
s = "monika"

print(s.__getattribute__("upper"))
```

6.  __dir__()

Use: Shows the attributes and methods available for an object.

```python
s = "monika"

print(s.__dir__())
```

7. __format__()

Use: Formats a string according to a specified format.

```python
s = "monika"

print(s.__format__(">10"))
```

8. __eq__()

Use: Compares two strings to check whether they are equal.

```python
s1 = "monika"
s2 = "mona"

print(s1.__eq__(s2))
print(s1==s2)
```

9. __doc__

Use: Shows the documentation of the str class.

```python
print(str.__doc__)
```

10. __contains__()

Use: Checks whether a substring exists inside a string.

```python
s = "monika"

print(s.__contains__("ni"))
```

11. __class__

Use: Shows the class/type of the object.

```python
s = "monika"

print(s.__class__)
```

12. __add__()

Use: Joins two strings using string addition.

```python
s1 = "Moni"
s2 = "ka"

print(s1.__add__(s2))
```

13. replace()

Use: Replaces one character or substring with another.

```python
s = "I like Java"
print(s.replace("Java", "Python"))
```

14. join()
Use: Joins multiple strings into one string using a separator.

```python
s = ["I", "love", "Python"]
print(" ".join(s))
```

15. split()

Use: Splits a string into a list.

```python
s = "I love Python"
print(s.split())
```

16. strip()

Use: Removes spaces from the beginning and end of a string.

```python
s = "   monika   "
print(s.strip())
```

17. endswith()

Use: Checks whether a string ends with specified text.

```python
s = "monika"
print(s.endswith("ka"))
```

18. startswith()

Use: Checks whether a string starts with specified text.

```python
s = "monika"
print(s.startswith("mon"))
```

19. count()

Use: Counts how many times a character or substring occurs.

```python
s = "banana"
print(s.count("a"))
```

20. lower()

Use: Converts all letters to lowercase.

```python
s = "MONIKA"
print(s.lower())
```

21. upper()

Use: Converts all letters to uppercase.

```python
s="monika"
print(s.upper())
```
